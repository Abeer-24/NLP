```python

import spacy
from spacy import displacy
```

```python

nlp = spacy.load("en_core_web_sm")

text = "Abeer studies at NIET in Greater Noida. Google was founded by Larry Page and Sergey Brin in 1998."

doc = nlp(text)

print("Input Text:")
print(text)
```

```python

print("Named Entities:")

for ent in doc.ents:
    print(ent.text, "->", ent.label_)
```

```python

for ent in doc.ents:
    print(ent.text, "->", ent.label_, "->", spacy.explain(ent.label_))
```

```python

displacy.render(doc, style="ent", jupyter=True)
```
