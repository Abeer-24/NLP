```python
from transformers import MarianMTModel, MarianTokenizer
import torch
```

```python
model_name = "Helsinki-NLP/opus-mt-en-hi"

tokenizer = MarianTokenizer.from_pretrained(model_name)
model = MarianMTModel.from_pretrained(model_name)

print("MarianMT Model Loaded Successfully")
```

```python
sentences = [
    "Hello, how are you?",
    "Artificial intelligence is changing the world.",
    "India is a beautiful country.",
    "Machine learning helps computers learn from data.",
    "Education is important for everyone."
]

print("English Sentences:")
for sentence in sentences:
    print(sentence)
```

```python
def translate(text):
    inputs = tokenizer(
        text,
        return_tensors="pt",
        padding=True,
        truncation=True
    )

    with torch.no_grad():
        translated = model.generate(
            **inputs,
            max_new_tokens=100
        )

    return tokenizer.decode(
        translated[0],
        skip_special_tokens=True
    )
```

```python
print("English to Hindi Translation:\n")

for sentence in sentences:
    hindi_text = translate(sentence)

    print("English:", sentence)
    print("Hindi:", hindi_text)
    print()
```

```python
user_text = input("Enter an English sentence: ")

translated_text = translate(user_text)

print("\nOriginal English:")
print(user_text)

print("\nTranslated Hindi:")
print(translated_text)
```

```python
import pandas as pd

results = []

for sentence in sentences:
    results.append({
        "English": sentence,
        "Hindi": translate(sentence)
    })

df = pd.DataFrame(results)

display(df)
```
