```python

import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python

data = {
    "text": [
        "I love this movie",
        "This movie is amazing",
        "What a wonderful experience",
        "I really enjoyed the film",
        "The movie was fantastic",
        "I hate this movie",
        "This film is terrible",
        "Very bad experience",
        "I disliked the movie",
        "The movie was boring"
    ],
    "label": [
        "Positive", "Positive", "Positive", "Positive", "Positive",
        "Negative", "Negative", "Negative", "Negative", "Negative"
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python

X = df["text"]
y = df["label"]

print("Text Data:")
print(X)

print("\nLabels:")
print(y)
```

```python

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.3,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python

tfidf = TfidfVectorizer()

X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

print("TF-IDF Training Shape:", X_train_tfidf.shape)
print("TF-IDF Testing Shape:", X_test_tfidf.shape)
```

```python

nb_model = MultinomialNB()

nb_model.fit(X_train_tfidf, y_train)

nb_predictions = nb_model.predict(X_test_tfidf)

print("Naive Bayes Accuracy:",
      accuracy_score(y_test, nb_predictions))
```

```python

svm_model = LinearSVC(random_state=42)

svm_model.fit(X_train_tfidf, y_train)

svm_predictions = svm_model.predict(X_test_tfidf)

print("SVM Accuracy:",
      accuracy_score(y_test, svm_predictions))
```

```python

print("Naive Bayes Classification Report:")
print(classification_report(y_test, nb_predictions, zero_division=0))

print("\nSVM Classification Report:")
print(classification_report(y_test, svm_predictions, zero_division=0))
```

```python

results = pd.DataFrame({
    "Model": ["Naive Bayes", "SVM"],
    "Accuracy": [
        accuracy_score(y_test, nb_predictions),
        accuracy_score(y_test, svm_predictions)
    ]
})

print("Model Comparison:")
display(results)
```

```python

new_text = ["This movie is wonderful and amazing"]

new_tfidf = tfidf.transform(new_text)

print("Input:", new_text[0])
print("Naive Bayes Prediction:", nb_model.predict(new_tfidf)[0])
print("SVM Prediction:", svm_model.predict(new_tfidf)[0])
```

```python

print("Naive Bayes Confusion Matrix:")
print(confusion_matrix(y_test, nb_predictions))

print("\nSVM Confusion Matrix:")
print(confusion_matrix(y_test, svm_predictions))
```
