```python

import nltk
nltk.download('punkt')
from nltk.tokenize import sent_tokenize,word_tokenize
```

```python
text = "Natural Language Processing is Intresting. It is used in AI."
```

```python
nltk.download('punkt_tab')
sentences = sent_tokenize(text)
for sentence in sentences:
  print(sentence)
```

```python
nltk.download('punkt_tab')
words = word_tokenize(text)
for word in words:
  print(word)
```

```python

import spacy
nlp = spacy.load('en_core_web_sm')
```

```python
doc = nlp(text)
for sent in doc.sents:
  print(sent.text)
```

```python

for token in doc:
  print(token.text)
```
