```python

from nltk.util import ngrams
from nltk.tokenize import word_tokenize
import nltk
```

```python

nltk.download('punkt')
nltk.download('punkt_tab')

```

```python

corpus = """Natural language processing is interesting.
Natural language processing helps computers understand text."""

print("Corpus:")
print(corpus)
```

```python

words = word_tokenize(corpus.lower())

print("Tokens:")
print(words)
```

```python

unigrams = list(ngrams(words, 1))

print("Unigrams:")
print(unigrams)
```

```python

bigrams = list(ngrams(words, 2))

print("Bigrams:")
print(bigrams)
```

```python

trigrams = list(ngrams(words, 3))

print("Trigrams:")
print(trigrams)
```

```python

print("Unigrams:", len(unigrams))
print(unigrams)

print("\nBigrams:", len(bigrams))
print(bigrams)

print("\nTrigrams:", len(trigrams))
print(trigrams)
```
