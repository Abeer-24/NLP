```python

from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
import pandas as pd
```

```python

documents = [
    "I love machine learning",
    "I love deep learning",
    "Machine learning is interesting"
]

print("Sample Documents:")
for doc in documents:
    print(doc)
```

```python

bow_vectorizer = CountVectorizer()
bow_matrix = bow_vectorizer.fit_transform(documents)

bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=bow_vectorizer.get_feature_names_out()
)

print("Bag-of-Words Representation:")
display(bow_df)
```

```python

tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(documents)

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf_vectorizer.get_feature_names_out()
)

print("TF-IDF Representation:")
display(tfidf_df.round(3))
```

```python

print("BoW Vocabulary:")
print(bow_vectorizer.get_feature_names_out())

print("\nTF-IDF Vocabulary:")
print(tfidf_vectorizer.get_feature_names_out())
```

```python

comparison = pd.DataFrame({
    "Word": bow_vectorizer.get_feature_names_out(),
    "BoW Count": bow_matrix.toarray()[0],
    "TF-IDF Score": tfidf_matrix.toarray()[0]
})

print("Comparison for Document 1:")
display(comparison.round(3))
```

```python

print("BoW vs TF-IDF")

print("BoW:")
print("Counts the frequency of each word in a document.")

print("\nTF-IDF:")
print("Assigns weights based on word frequency and how informative")
print("a word is across all documents.")
```
