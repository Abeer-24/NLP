```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
documents = [
    "Machine learning is a branch of artificial intelligence.",
    "Deep learning uses neural networks for complex tasks.",
    "Python is a popular programming language.",
    "Artificial intelligence is transforming healthcare.",
    "Data science involves statistics and machine learning.",
    "Natural language processing helps computers understand text.",
    "Computer networks connect multiple devices together.",
    "Deep learning and machine learning are important AI technologies."
]

doc_names = [f"Document {i+1}" for i in range(len(documents))]
```

```python
for name, doc in zip(doc_names, documents):
    print(name, ":", doc)
```

```python
vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(documents)

print("TF-IDF Matrix Shape:", tfidf_matrix.shape)
print("\nVocabulary:")
print(vectorizer.get_feature_names_out())
```

```python
query = input("Enter your search query: ")

print("Search Query:", query)
```

```python
query_vector = vectorizer.transform([query])

print("Query Vector:")
print(query_vector.toarray())
```

```python
similarity_scores = cosine_similarity(
    query_vector,
    tfidf_matrix
).flatten()

print("Similarity Scores:")
print(similarity_scores)
```

```python
ranked_indices = similarity_scores.argsort()[::-1]

print("Ranked Search Results:\n")

for rank, index in enumerate(ranked_indices, start=1):
    if similarity_scores[index] > 0:
        print(f"Rank {rank}")
        print("Document:", doc_names[index])
        print("Similarity Score:", round(similarity_scores[index], 4))
        print("Content:", documents[index])
        print()
```

```python
results = pd.DataFrame({
    "Document": doc_names,
    "Content": documents,
    "Similarity Score": similarity_scores
})

results = results.sort_values(
    by="Similarity Score",
    ascending=False
)

display(results)
```

```python
new_query = input("Enter another search query: ")

new_query_vector = vectorizer.transform([new_query])

new_scores = cosine_similarity(
    new_query_vector,
    tfidf_matrix
).flatten()

new_ranking = new_scores.argsort()[::-1]

print("\nTop Search Results:\n")

for index in new_ranking:
    if new_scores[index] > 0:
        print(doc_names[index], "-", documents[index])
        print("Score:", round(new_scores[index], 4))
        print()
```
