```python
import spacy

!python -m spacy download en_core_web_sm

nlp = spacy.load("en_core_web_sm")

print("spaCy loaded successfully!")
```

```python
text = """
Rahul bought a laptop yesterday.
He uses it for his college assignments.
Rahul says the laptop is very useful.
"""

doc = nlp(text)

print("Input Text:")
print(text)
```

```python
pronoun_references = {
    "he": "Rahul",
    "him": "Rahul",
    "his": "Rahul",
    "she": "Female person (antecedent needed)",
    "her": "Female person (antecedent needed)",
    "it": "Previously mentioned object (antecedent needed)",
    "they": "Previously mentioned plural entity (antecedent needed)"
}

print("Detected Pronouns and Possible References:")

for token in doc:
    if token.text.lower() in pronoun_references:
        print(token.text, "->", pronoun_references[token.text.lower()])
```

```python
print("Named Entities:")

for ent in doc.ents:
    print(ent.text, "->", ent.label_)

print("\nSentences:")

for sent in doc.sents:
    print(sent.text)
```
