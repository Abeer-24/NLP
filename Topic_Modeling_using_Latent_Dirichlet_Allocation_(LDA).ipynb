```python
import nltk
import gensim
from gensim import corpora
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
documents = [
    "Machine learning and artificial intelligence are transforming technology.",
    "Deep learning uses neural networks to solve complex problems.",
    "Python is a popular programming language for machine learning.",
    "Football and cricket are popular sports around the world.",
    "Players practice daily to improve their sports performance.",
    "Cricket tournaments attract millions of sports fans.",
    "Artificial intelligence is used in healthcare and education.",
    "Neural networks are an important part of deep learning."
]
```

```python
stop_words = set(stopwords.words('english'))

processed_docs = []

for doc in documents:
    words = word_tokenize(doc.lower())
    words = [
        word for word in words
        if word.isalpha() and word not in stop_words
    ]
    processed_docs.append(words)

print(processed_docs)
```

```python
dictionary = corpora.Dictionary(processed_docs)

print("Word Dictionary:")
print(dictionary.token2id)
```

```python
corpus = [
    dictionary.doc2bow(doc)
    for doc in processed_docs
]

print("Bag of Words Corpus:")
print(corpus)
```

```python
lda_model = gensim.models.LdaModel(
    corpus=corpus,
    id2word=dictionary,
    num_topics=2,
    random_state=42,
    passes=20
)
```

```python
print("Topics discovered by LDA:\n")

for topic in lda_model.print_topics():
    print(topic)
```

```python
for i, doc in enumerate(corpus):
    topics = lda_model.get_document_topics(doc)

    print("\nDocument:", documents[i])
    print("Topic Distribution:", topics)
```

```python
for topic_id in range(2):
    print(f"\nTopic {topic_id + 1}:")
    words = lda_model.show_topic(topic_id, topn=5)

    for word, probability in words:
        print(word, ":", round(probability, 3))
```
