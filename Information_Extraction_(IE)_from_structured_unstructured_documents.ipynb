```python
import pandas as pd
import spacy
import re

nlp = spacy.load("en_core_web_sm")
```

```python
data = {
    "Name": ["Rahul Sharma", "Priya Verma", "Amit Kumar"],
    "Email": [
        "rahul@gmail.com",
        "priya@gmail.com",
        "amit@gmail.com"
    ],
    "City": ["Delhi", "Mumbai", "Noida"],
    "Course": ["CSE", "AI", "IT"]
}

df = pd.DataFrame(data)

print("Structured Document:")
display(df)
```

```python
print("Names:")
print(df["Name"].tolist())

print("\nEmail Addresses:")
print(df["Email"].tolist())

print("\nCities:")
print(df["City"].tolist())

print("\nCourses:")
print(df["Course"].tolist())
```

```python
text = """
Rahul Sharma works at Google in Delhi.
He joined the company on 15 January 2024.
His email is rahul@gmail.com and phone number is 9876543210.

Priya Verma works at Microsoft in Mumbai.
She joined on 20 March 2023.
Her email is priya@gmail.com.
"""
print(text)
```

```python
doc = nlp(text)

for ent in doc.ents:
    print(ent.text, ":", ent.label_)
```

```python
emails = re.findall(
    r'[\w.-]+@[\w.-]+\.\w+',
    text
)

print("Extracted Emails:")
print(emails)
```

```python
phone_numbers = re.findall(
    r'\b\d{10}\b',
    text
)

print("Extracted Phone Numbers:")
print(phone_numbers)
```

```python
extracted_data = {
    "Emails": emails,
    "Phone Numbers": phone_numbers
}

print("Extracted Information:")
display(pd.DataFrame({
    "Emails": pd.Series(emails),
    "Phone Numbers": pd.Series(phone_numbers)
}))
```

```python
print("Noun Phrases:")

for chunk in doc.noun_chunks:
    print(chunk.text)

print("\nImportant Keywords:")

for token in doc:
    if token.pos_ in ["NOUN", "PROPN"] and not token.is_stop:
        print(token.text)
```

```python
print("INFORMATION EXTRACTION SUMMARY")
print("--------------------------------")

print("Names and Organizations:")
for ent in doc.ents:
    if ent.label_ in ["PERSON", "ORG"]:
        print(ent.text, "-", ent.label_)

print("\nEmails:", emails)
print("Phone Numbers:", phone_numbers)
```
