```python

import nltk
import numpy as np
from gensim.models import Word2Vec
from nltk.tokenize import word_tokenize
```

```python

nltk.download('punkt')
nltk.download('punkt_tab')
```

```python

doc1 = "Machine learning is useful for data analysis"
doc2 = "Artificial intelligence helps computers learn"
doc3 = "I enjoy playing cricket and football"

print("Document 1:", doc1)
print("Document 2:", doc2)
print("Document 3:", doc3)
```

```python

documents = [doc1, doc2, doc3]

tokenized_docs = [
    word_tokenize(doc.lower())
    for doc in documents
]

print("Tokenized Documents:")
for doc in tokenized_docs:
    print(doc)
```

```python

model = Word2Vec(
    sentences=tokenized_docs,
    vector_size=100,
    window=5,
    min_count=1,
    workers=1,
    seed=42,
    epochs=100
)

print("Word2Vec model trained successfully!")
```

```python

distance_12 = model.wv.wmdistance(tokenized_docs[0], tokenized_docs[1])
distance_13 = model.wv.wmdistance(tokenized_docs[0], tokenized_docs[2])
distance_23 = model.wv.wmdistance(tokenized_docs[1], tokenized_docs[2])

print("WMD between Document 1 and Document 2:", distance_12)
print("WMD between Document 1 and Document 3:", distance_13)
print("WMD between Document 2 and Document 3:", distance_23)
```

```python

pairs = [
    ("Document 1 and Document 2", distance_12),
    ("Document 1 and Document 3", distance_13),
    ("Document 2 and Document 3", distance_23)
]

for name, distance in pairs:
    print(name, "-> Distance:", round(distance, 4))
```
