```python

from gensim.models import Word2Vec
from nltk.tokenize import word_tokenize
import nltk
```

```python

nltk.download('punkt')
nltk.download('punkt_tab')
```

```python

corpus = [
    "Machine learning is a part of artificial intelligence",
    "Deep learning uses neural networks",
    "Artificial intelligence makes machines intelligent",
    "Machine learning uses data for prediction",
    "Neural networks are used in deep learning"
]

print("Custom Corpus:")
for sentence in corpus:
    print(sentence)
```

```python

tokenized_corpus = [
    word_tokenize(sentence.lower())
    for sentence in corpus
]

print("Tokenized Corpus:")
print(tokenized_corpus)
```

```python

model = Word2Vec(
    sentences=tokenized_corpus,
    vector_size=100,
    window=5,
    min_count=1,
    workers=1,
    sg=1,
    seed=42,
    epochs=100
)

print("Word2Vec model trained successfully!")
```

```python

word = "learning"

print("Embedding for:", word)
print(model.wv[word])
```

```python

word = "learning"

print("Words similar to:", word)
print(model.wv.most_similar(word, topn=3))
```

```python

print("Vocabulary:")
print(list(model.wv.index_to_key))
```

```python

print("Vector size:", model.wv.vector_size)
print("Vocabulary size:", len(model.wv))
```
