```python
import pandas as pd
import nltk
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report
```

```python
nltk.download('stopwords')
```

```python
data = {
    "Review": [
        "This phone has an excellent camera and battery life.",
        "The product is amazing and works perfectly.",
        "I love this phone. It is worth the money.",
        "Very good quality and fast performance.",
        "The delivery was quick and the service was excellent.",
        "This product is very bad and disappointing.",
        "The battery drains quickly and camera quality is poor.",
        "I hate this product. It is completely useless.",
        "Worst service and poor customer support.",
        "The product stopped working after two days."
    ],
    "Sentiment": [
        "Positive", "Positive", "Positive", "Positive", "Positive",
        "Negative", "Negative", "Negative", "Negative", "Negative"
    ]
}

df = pd.DataFrame(data)
display(df)
```

```python
stop_words = set(stopwords.words('english'))

def preprocess(text):
    words = text.lower().split()
    words = [
        word for word in words
        if word not in stop_words
    ]
    return " ".join(words)

df["Cleaned_Review"] = df["Review"].apply(preprocess)

display(df[["Review", "Cleaned_Review", "Sentiment"]])
```

```python
vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(df["Cleaned_Review"])
y = df["Sentiment"]

print("TF-IDF Matrix Shape:", X.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.3,
    random_state=42,
    stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python
model = MultinomialNB()

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("Predictions:", y_pred)
```

```python
model = MultinomialNB()

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("Predictions:", y_pred)
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))
```

```python
new_review = ["The phone has excellent performance and amazing battery life."]

cleaned_review = [preprocess(new_review[0])]

review_vector = vectorizer.transform(cleaned_review)

prediction = model.predict(review_vector)

print("Review:", new_review[0])
print("Predicted Opinion:", prediction[0])
```

```python
new_reviews = [
    "The product is excellent and easy to use.",
    "Very poor quality and terrible service.",
    "I am happy with this purchase."
]

review_vectors = vectorizer.transform(
    [preprocess(review) for review in new_reviews]
)

predictions = model.predict(review_vectors)

results = pd.DataFrame({
    "Review": new_reviews,
    "Predicted Sentiment": predictions
})

display(results)
```
